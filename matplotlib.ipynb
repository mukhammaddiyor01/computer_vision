```python
import matplotlib.pyplot as plt
import numpy as np
```

# Basic Line Plot

```python
x = np.array([1, 2, 3, 4])
y = np.array([10, 22, 12, 30])

plt.plot(x,y)
plt.title("Basic Line Plot")
plt.xlabel("x values")
plt.ylabel("y values")
plt.show()
```

# Line styles and colors

```python
x = np.array([1, 2, 3, 4])
y = np.array([3, 8, 1, 12])

plt.plot(y, linestyle='dotted', color='red', marker='o')
plt.title("Styled Line Example")
plt.show()
```

# Multiple times in One Graph with Numpy andMatplotlib

```python
x = np.array([1, 2, 3, 4])
y1 = np.array([10, 22, 12, 30])
y2 = np.array([3, 7, 6, 9])

plt.plot(x, y1, label='Line 1', color='red', marker='s')
plt.plot(x, y2, label='Line 2', color='blue', linestyle='--', marker='o')
plt.title("Multiple Line Examples")
plt.xlabel("x values")
plt.ylabel("y values")
plt.legend()
plt.show()
```

Bar Chart with Numpy and Matplotlib

```python
x = np.array(["Salary", "Expense", "Charity", "Donation"])
y = np.array([3, 8, 1, 10])

plt.bar(x,y, color='orange')
plt.title("Bar Chart Example")
plt.xlabel('Categories')
plt.ylabel('Values')
plt.show()
```

Scatter Plot with Numpy and Matplotlib

```python
x = np.array([1, 2, 3, 4, 5])
y = np.array([3, 7, 6, 9, 8])

plt.scatter(x,y, color="purple")
plt.title("Scatter Plot Example")
plt.xlabel('x values')
plt.ylabel('y values')
plt.show()
```

Histogram with Numpy and Matplotlib with AI.

```python
data = np.array([3, 4, 5, 7, 6, 8, 9, 10])

plt.hist(data, bins=5, color='green', edgecolor='black')
plt.title("Histogram Example")
plt.xlabel('Values')
plt.ylabel('count')
plt.show()
```

Subplot(multiple Graphs Together)

```python
x = np.array([1, 2, 3, 4])
y1 = np.array([10, 22, 12, 30])
y2 = np.array([3, 7, 6, 9])

plt.figure(figsize=(10, 5)) # width(wide) | 5 => height(tall)

plt.subplot(1, 2, 1)
plt.plot(x, y1, color='red')
plt.title("Plot 1")

plt.subplot(1, 2, 2)
plt.bar(x, y2, color='blue')
plt.title("Plot 2")
plt.tight_layout()

plt.show()
```

```python
x = np.array([1, 2, 3, 4])
y = np.array([3, 8, 1, 10])

plt.plot(x, y, color='green')
plt.title("Save Figure Example")

plt.savefig('plot.png') # saves in current folder
plt.show()
```

Plot_wireframe

```python
import matplotlib.pyplot as plt

from mpl_toolkits.mplot3d import axes3d

plt.style.use('_mpl-gallery')

# Make data
X, Y, Z = axes3d.get_test_data(0.05)

# Plot
fig, ax = plt.subplots(subplot_kw={"projection": "3d"})
ax.plot_wireframe(X, Y, Z, rstride=10, cstride=10)

ax.set(xticklabels=[],
       yticklabels=[],
       zticklabels=[])

plt.show()
```

Scatters

```python
import matplotlib.pyplot as plt
import numpy as np

plt.style.use('_mpl-gallery')

# Make data
np.random.seed(19680801)
n = 100
rng = np.random.default_rng()
xs = rng.uniform(23, 32, n)
ys = rng.uniform(0, 100, n)
zs = rng.uniform(-50, -25, n)

# Plot
fig, ax = plt.subplots(subplot_kw={"projection": "3d"})
ax.scatter(xs, ys, zs)

ax.set(xticklabels=[],
       yticklabels=[],
       zticklabels=[])

plt.show()
```
